```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("Mall_Customers.csv")
data.head()
```

```python
X = data[['Annual Income (k$)', 'Spending Score (1-100)']].values
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
y_kmeans = kmeans.fit_predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], s=200, c='red')
plt.title("K-Means Clustering")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.show()
```

```python
def weighted_kmeans(X, weights, k=5, max_iters=100):
    n = X.shape[0]
    centroids = X[np.random.choice(n, k, replace=False)]
    
    for _ in range(max_iters):
        clusters = [[] for _ in range(k)]
        
        for i in range(n):
            distances = [np.linalg.norm(X[i] - c) for c in centroids]
            cluster_idx = np.argmin(distances)
            clusters[cluster_idx].append(i)
        
        new_centroids = []
        
        for j in range(k):
            if len(clusters[j]) == 0:
                new_centroids.append(centroids[j])
                continue
            
            weighted_sum = np.sum([weights[i] * X[i] for i in clusters[j]], axis=0)
            total_weight = np.sum([weights[i] for i in clusters[j]])
            
            new_centroid = weighted_sum / total_weight
            new_centroids.append(new_centroid)
        
        new_centroids = np.array(new_centroids)
        
        if np.allclose(centroids, new_centroids):
            break
        
        centroids = new_centroids
    
    labels = np.zeros(n)
    
    for i in range(n):
        distances = [np.linalg.norm(X[i] - c) for c in centroids]
        labels[i] = np.argmin(distances)
    
    return labels, centroids
```

```python
weights = X[:,1] / np.max(X[:,1])
```

```python
labels_wk, centroids_wk = weighted_kmeans(X, weights, k=5)
```

```python
plt.scatter(X[:,0], X[:,1], c=labels_wk, cmap='plasma')
plt.scatter(centroids_wk[:,0], centroids_wk[:,1], s=200, c='black')
plt.title("Weighted K-Means Clustering")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.show()
```
